<p><b>What is a Context Manager?</b>
<p>A Context Manager in Python is an object that manages resources automatically. It is used to allocate and release resources properly.
<p>It helps ensure that resources such as files, database connections, and locks are handled safely, even if an error occurs.
<p>Python uses the with statement to work with context managers.
<p>In short:A context manager is a Python object that manages the setup and cleanup of resources automatically using the with statement.


<p><b>Syntax of Context Manager</b></p>
<p><b>with expression as variable:</b></p> 
  &emsp;statements</b>
<p>with – Starts the context management block.
<p>expression – Creates or returns a context manager.
<p>as – Stores the returned resource in a variable.
<p>statements – Operations performed using the resource.


```python
#Example: Reading a File
with open("sample.txt", "r") as file:
    data = file.read()
    print(data)
```

<p><b>Explanation:</b>
<p>1.open() opens the file.
<p>2.The with statement manages the file.
<p>3.file.read() reads the contents.
<p>4.After the block finishes, Python automatically closes the file.
    
<p>Note: The programmer does not need to call file.close() explicitly.

```python
#Why Do We Use Context Managers?
#Consider opening a file without a context manager.
#Without Context Manager:

file = open("sample.txt", "r")
data = file.read()
file.close()
#Here, we must manually close the file.
#With Context Manager:

with open("sample.txt", "r") as file:
    data = file.read()
#Here, Python automatically closes the file when the block exits, including when an exception occurs.
```

```python
#Types of Context Managers
#A. Using the with Statement
#Python's built-in functions, such as open(), support context management.
with open("test.txt", "w") as f:
    f.write("Hello Python")
```

```python
#B. Using Class-Based Context Managers
#A class can become a context manager by implementing two special methods:
#__enter__() – Executes when entering the with block.
#__exit__() – Executes when leaving the with block.
```

```python
class MyContext:
    def __enter__(self):
        print("Entering the context")
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        print("Exiting the context")

with MyContext():
    print("Inside the context")
```

<p>Explanation:
<p>1.__enter__() executes first.
<p>2.The statements inside the with block execute.
<p>3.__exit__() executes automatically after the block finishes.


```python
#Context Manager Using contextlib
#Python provides a module called contextlib to create context managers easily.
#It provides the @contextmanager decorator.
from contextlib import contextmanager
@contextmanager
def my_context():
    print("Resource opened")

    try:
        yield
    finally:
        print("Resource closed")

with my_context():
    print("Working with resource")
```

```python
#Here:
#•	Code before yield performs setup.
#•	Code inside with performs the main operation.
#•	Code inside finally performs cleanup.
```

<p><b>Advantages of Context Managers</b>
<p>1.	Automatically manages resources.
<p>2.	Reduces the chance of resource leaks.
<p>3.	Makes code cleaner and easier to read.
<p>4.	Handles cleanup even when exceptions occur.
<p>5.	Reduces the need for manual cleanup.
<p>6.	Useful for file handling and database connections.
<p><b>Applications of Context Managers</b>
<p>File handling.
<p>Database connections.
<p>Lock management in multithreading.
<p>Network connections.
<p>Resource management.

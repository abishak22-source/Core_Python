<p><b>Lambda Expressions in Python</b>
<p><b>1. What is a Lambda Expression?</b>
<p>A Lambda Expression in Python is a small anonymous function (a function without a name) that can contain only one expression.
<p>It is used to create simple functions in a single line without using the def keyword.
<p><b>Syntax:</b> lambda arguments: expression
<p>lambda – Keyword used to create a lambda function.
<p>arguments – Input values passed to the function.
<p>expression – Operation performed by the function.

```python
#Simple Example
add = lambda a, b: a + b
print(add(10, 20))
```

Explanation:

    lambda a, b takes two arguments.

    a + b performs addition.

    add stores the lambda function.

    add(10, 20) calls the function.

```python
#3. Examples of Lambda Expressions
#Example 1: Square of a Number
square = lambda x: x * x
print(square(5))
```

```python
#Example 2: Find the Cube
cube = lambda x: x ** 3
print(cube(3))
```

```python
#Example 3: Find the Maximum of Two Numbers
maximum = lambda a, b: a if a > b else b
print(maximum(10, 20))
```

```python
#Example 4: Check Even or Odd
check = lambda x: "Even" if x % 2 == 0 else "Odd"
print(check(8))
```

```python
#4. Lambda Function with map()
#The map() function applies a given function to every element in a sequence.
numbers = [1, 2, 3, 4, 5]

result = list(map(lambda x: x * 2, numbers))

print(result)
```

```python
#5. Lambda Function with filter()
#The filter() function selects elements that satisfy a given condition.
numbers = [1, 2, 3, 4, 5, 6]
result = list(filter(lambda x: x % 2 == 0, numbers))
print(result)
```

```python
#6. Lambda Function with reduce()

#The reduce() function applies a function cumulatively to the elements of a sequence to produce a single result.

#It is available in the functools module.
from functools import reduce

numbers = [1, 2, 3, 4, 5]

result = reduce(lambda a, b: a + b, numbers)

print(result)
```

<p><b>Advantages of Lambda Expressions</b>

 <p>   Reduces the amount of code.

   <p>  Makes simple functions easier to write.

   <p>  Does not require the def keyword.

   <p>  Useful with map(), filter() and reduce().

  <p>   Useful for short operations and sorting.

<p><b> Limitations of Lambda Expressions</b>

  <p>   Can contain only one expression.

  <p>   Difficult to understand when the logic is complex.

   <p>  Not suitable for large functions.

  <p>   Cannot contain statements such as return, for or while as normal statements.

<p><b>Important Points</b>

 <p>   Lambda functions are also called anonymous functions.

  <p>  Lambda functions are created using the lambda keyword.

 <p>   A lambda function can accept multiple arguments.

 <p>   A lambda function contains only one expression.

 <p>   The result is returned automatically.

  <p>  Lambda functions are commonly used with map(), filter() and reduce().

```python
#Sets in Python

#A set in Python is a collection of unique elements. It is unordered, mutable, and does not allow duplicate values.
#1. Creating a set
# Creating a set
my_set = {10, 20, 30, 40}

print(my_set)
```

```python
#2. Removing duplicate values
numbers = {10, 20, 10, 30, 20}

print(numbers)
```

```python
#3. Adding elements:Use add() to add one element.
s = {1, 2, 3}
s.add(4)

print(s)
```

```python
#4. Removing elements
s = {10, 20, 30, 40}
s.remove(20)  # Removes 20
s.discard(50) # No error if element is absent
print(s)
```

```python
#5. Set operations
#Consider two sets:
#A = {1, 2, 3, 4}
#B = {3, 4, 5, 6}
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
print("UNION",A | B)   #Union (|) – Combines all elements.
print("INTERSECTION",A & B)   #Intersection (&) – Returns common elements.
print("DIFFERENCE",A - B)   #Difference (-) – Returns elements in A but not in B.
print("Symmetric Difference",A ^ B)   #Symmetric Difference (^) – Returns elements that are not common.
```

Method -------- Description

add()	Adds an element

update()	Adds multiple elements

remove()	Removes an element; error if absent

discard()	Removes an element without error if absent

pop()	Removes an arbitrary element

clear()	Removes all elements

copy()	Creates a copy

issubset()	Checks whether a set is a subset

issuperset()	Checks whether a set is a superset

isdisjoint()	Checks whether two sets have no common elements

```python
#Empty set
a = set()  # Correct way
b = {}     # Creates an empty dictionary

print(type(a))
print(type(b))
```

```python
# Checking membership
fruits = {"apple", "banana", "orange"}

print("apple" in fruits)
print("grapes" not in fruits)
```

```python
#Set comprehension
squares = {x*x for x in range(1, 6)}

print(squares)
```

Important points to remember:

    Sets do not allow duplicate elements.

    Sets are unordered.

    Sets are mutable.

    Set elements must be immutable (such as numbers, strings, or tuples containing immutable values).

    Sets do not support indexing or slicing.

<font color="red" size="6">NAME SPACE AND SCOPE</font>

<p>In Python, Namespaces and Scopes are important concepts used to manage variables, functions, and objects in a program.
<p>They help Python identify where a variable is created and where it can be accessed.
<p><b>Namespace: A namespace is a collection of names and their corresponding objects.
<p><b>Scope: Scope refers to the region of a program where a variable or name can be accessed.


```python
#A namespace is a system that stores names and the objects associated with them.
a = 10
b = 20

def display():
    print("Hello Python")

print(a)
print(b)
display()
```

<p>Explanation:
<p>•	a and b are variable names.
<p>•	display is a function name.
<p>•	Python stores these names in a namespace and maps them to their corresponding objects.


```python
#Types of Namespaces in Python
#1. Built-in Namespace
#•	Contains predefined names provided by Python.
#•	These names can be used anywhere in a program.
#•	Examples: print(), len(), input(), int().
numbers = [10, 20, 30]

print(len(numbers))
```

```python
#Here, print() and len() belong to the built-in namespace.
```

```python
#2.Global Namespace
#•	Contains variables and functions defined at the main program level.
#•	These variables can be accessed throughout the module
x = 100

def display():
    print("Global value:", x)

display()
print(x)
```

```python
#Here, x is a global variable because it is defined outside the function.
```

```python
#3. Enclosing Namespace
#•	Exists in nested functions.
#•	Contains variables defined in the outer function.
#•	These variables can be accessed by the inner function.
def outer():
    x = 50

    def inner():
        print("Enclosing value:", x)

    inner()

outer()
```

```python
#Here, x belongs to the enclosing namespace of the inner function
```

```python
#4. Local Namespace
#•	Contains variables defined inside a function.
#•	These variables are normally accessible only within that function.
def display():
    x = 10
    print("Local value:", x)

display()
```

```python
#Here, x is a local variable because it is declared inside the function.
```

<p><b>What is Scope?</b>
<p>Scope determines where a variable can be accessed in a Python program.
<p>Python follows the <b>LEGB rule</b> to search for variable names.
<p><b>Types of Scope in Python (LEGB Rule):</b>
<p><b>L – Local</b>: Inside the current function
<p><b>E – Enclosing</b>: Inside the outer function
<p><b>G – Global</b>: At the module level
<p><b>B – Built-in</b>: Predefined Python names


```python
#Example of LEGB Rule
x = "Global"

def outer():
    x = "Enclosing"

    def inner():
        x = "Local"
        print(x)

    inner()

outer()
print(x)
```

<p>Explanation:
<p>•	The inner function prints the local variable x.
<p>•	The outer function has an enclosing variable x.
<p>•	The main program has a global variable x.
<p>•	Python uses the nearest matching variable.


```python
#Global Keyword:The global keyword is used to modify a global variable inside a function.
#Example:
x = 10

def display():
    global x
    x = 20

display()

print(x)
```

```python
#Explanation: The global keyword allows the function to modify the global variable instead of creating a new local variable.
```

```python
#Nonlocal Keyword:The nonlocal keyword is used to modify a variable in the enclosing function.
def outer():
    x = 10

    def inner():
        nonlocal x
        x = 20

    inner()
    print(x)

outer()
```

```python
#Explanation: The nonlocal keyword modifies the variable x in the outer function.
```

<p>Namespaces and scopes are important concepts in Python. A namespace stores names and their corresponding objects, whereas scope determines where those names can be accessed. Python follows the LEGB rule (Local, Enclosing, Global, Built-in) to find variable names.

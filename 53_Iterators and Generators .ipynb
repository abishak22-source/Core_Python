<p><b>What is an Iterator?</b>
<p>An Iterator is an object that allows us to access elements of a collection one at a time.
<p>It is used to traverse elements such as lists, tuples, strings and dictionaries.
<p><b>In short:</b>An iterator is an object that implements the __iter__() and __next__() methods to access elements one by one.


```python
#Example of Iterator
numbers = [10, 20, 30, 40]
it = iter(numbers)
print(next(it))
print(next(it))
print(next(it))
print(next(it))
```

<p><b>Explanation:</b>
<p>•	iter() creates an iterator from an iterable.
<p>•	next() returns the next element.
<p>•	Each call to next() moves to the next element.


<p><b>Important Functions of Iterator</b>
<p><b>iter():</b>Creates an iterator
<p><b>next():</b>Returns the next element
<p><b>__iter__():</b>Returns the iterator object
<p><b>__next__():</b>Returns the next value


```python
fruits = ["Apple", "Banana", "Mango"]

for fruit in fruits:
    print(fruit)
```

```python
#The for loop internally uses the iterator protocol to access elements.
```

```python
#StopIteration Exception
#When an iterator has no more elements, Python raises the StopIteration exception.
numbers = [1, 2]

it = iter(numbers)

print(next(it))
print(next(it))
print(next(it))
```

```python
#The third next() call raises an exception because there are no more elements.
```

<p><b>Generators in Python</b>
<p>
A Generator is a special type of iterator that generates values one at a time instead of storing all values in memory.
<p>Generators are created using:
<p><b>yield keyword.</b>
<p><b>Generator expressions.</b>
<p>In short:A generator is a special function that uses the yield keyword to return values one at a time and remembers its execution state between calls.


```python
def my_generator():
    yield 10
    yield 20
    yield 30

gen = my_generator()

print(next(gen))
print(next(gen))
print(next(gen))
```

<p>Explanation:
<p>•	yield returns a value.
<p>•	The function pauses after each yield.
<p>•	When next() is called again, execution resumes from where it stopped.


```python
#Example Using yield
def count():
    yield 1
    yield 2
    yield 3

for value in count():
    print(value)
```

<p><b>Generator Expression</b>
<p>A generator expression is a short way of creating a generator.
<p><b>Syntax:</b> (expression for item in iterable)


```python
numbers = (x * x for x in range(1, 6))

for num in numbers:
    print(num)
```

```python
#Notice the difference:
#•	List comprehension uses square brackets [].
#•	Generator expression uses parentheses ().
```

```python
#Example: List Comprehension
numbers = (x * x for x in range(1, 6))
# List comprehension
a = [x*x for x in range(5)]

# Generator expression
b = (x*x for x in range(5))
print("A number is",a)
for b in b:
    print("B number is",b)
```

<p><b>Advantages of Generators</b>
<p>1.	Saves memory because values are generated one at a time.
<p>2.	Useful for processing large amounts of data.
<p>3.	Makes code simple and readable.
<p>4.	Supports lazy evaluation.
<p>5.	Maintains the execution state between yield statements.
<p><b>5. Advantages of Iterators</b>
<p>1.	Access elements one by one.
<p>2.	Supports sequential data traversal.
<p>3.	Useful with lists, tuples and strings.
<p>4.	Reduces the need for manual indexing.
<p>5.	Works with Python for loops.


```python
#Real-Time Example: Generating Even Numbers
def even_numbers(n):
    for i in range(1, n+1):
        if i % 2 == 0:
            yield i

for num in even_numbers(10):
    print(num)
```

<p><b>Explanation:</b>
<p>1.	The function checks numbers from 1 to 10.
<p>2.	It selects only even numbers.
<p>3.	yield generates each even number one at a time.
<p>4.	The for loop displays the generated values.
<p><b>Remember these three points:</b>
<p><b>Iterator</b> = Access elements one by one.
<p><b>Generator</b> = Produces values one at a time using yield.
<p><b>next()</b> = Retrieves the next available value.

```python
#Printing Patterns in Python
#Pattern printing in Python is a common programming exercise that helps you understand nested loops (for and while), rows, columns, and logic.
```

```python
#1. Square star pattern
for i in range(4):
    for j in range(4):
        print("*", end=" ")
    print()
```

```python
#2. Right-angled triangle
for i in range(1, 6):
    for j in range(i):
        print("*", end=" ")
    print()
```

```python
#3. Inverted right triangle
for i in range(5, 0, -1):
    for j in range(i):
        print("*", end=" ")
    print()
```

```python
#4. Right-aligned triangle
for i in range(1, 6):
    print("  " * (5-i), end="")
    print("* " * i)
```

```python
#5. Pyramid pattern
n = 5

for i in range(1, n+1):
    print(" " * (n-i) + "*" * (2*i-1))
```

```python
#6. Inverted pyramid
n = 5

for i in range(n, 0, -1):
    print(" " * (n-i) + "*" * (2*i-1))
```

```python
#7. Diamond pattern
n = 5

for i in range(1, n+1):
    print(" " * (n-i) + "*" * (2*i-1))

for i in range(n-1, 0, -1):
    print(" " * (n-i) + "*" * (2*i-1))
```

```python
#8. Number triangle
for i in range(1, 6):
    for j in range(1, i+1):
        print(j, end=" ")
    print()
```

```python
#9. Repeated number triangle
for i in range(1, 6):
    for j in range(i):
        print(i, end=" ")
    print()
```

```python
#10. Floyd's triangle
num = 1

for i in range(1, 6):
    for j in range(i):
        print(num, end=" ")
        num += 1
    print()
```

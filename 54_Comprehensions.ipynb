<font color="red" size="6">Comprehensions</font>

<p>Comprehensions in Python are a short and simple way to create new collections from existing iterables such as lists, tuples, dictionaries and sets.
<p>They help to write code in a single line instead of using lengthy loops.
<p>Comprehensions are concise Python expressions used to create new collections from existing iterables, with or without conditions.


<p>Types of Comprehensions in Python
<p><b>Python has four main types of comprehensions:</b>
<p><b>1.	List Comprehension:List comprehension is used to create a new list from an existing iterable.</b>
<p><b>2.	Set Comprehension:Set comprehension is used to create a set using an existing iterable.</b>
<p><b>3.	Dictionary Comprehension:Dictionary comprehension is used to create dictionaries in a single line.</b>
<p><b>4.	Generator Comprehension:Generator comprehension (also called a generator expression) creates a generator that produces values one at a time.</b>


<p>List Comprehension
<p><b>Syntax:</b>[expression for item in iterable if condition]


```python
#Example 1: Create a list of numbers
numbers = [x for x in range(1, 6)]
print(numbers)
```

```python
#Example 2: Find squares of numbers
squares = [x*x for x in range(1, 6)]
print(squares)
```

```python
#Example 3: Find even numbers
numbers = [1, 2, 3, 4, 5, 6]
even = [x for x in numbers if x % 2 == 0]
print(even)
```

<p>Explanation:
<p>for x in numbers accesses each element.
<p>if x % 2 == 0 checks whether the number is even.
<p>The result is stored in a new list.


<p>Set Comprehension:Set comprehension is used to create a set using an existing iterable.
<p><b>Syntax:</b>{expression for item in iterable if condition}


```python
numbers = [1, 2, 2, 3, 4, 4, 5]
result = {x for x in numbers}
print(result)
```

<p>Explanation:
<p>Curly braces {} are used.
<p>Duplicate elements are automatically removed.
<p>Sets do not guarantee a particular display order.

```python
#Example 2: Find even numbers
numbers = [1, 2, 3, 4, 5, 6]
even = {x for x in numbers if x % 2 == 0}
print(even)
```

<p>Dictionary Comprehension:Dictionary comprehension is used to create dictionaries in a single line.
<p><b>Syntax:</b>{key_expression: value_expression for item in iterable if condition}


```python
#Example 1: Create a dictionary
numbers = [1, 2, 3, 4, 5]
result = {x: x*x for x in numbers}
print(result)
```

<p>Explanation:
<p>x represents the dictionary key.
<p>x*x represents the dictionary value.
<p>Each number is stored with its square.


```python
#Example 2: Dictionary of even numbers
numbers = [1, 2, 3, 4, 5, 6]
result = {x: x*x for x in numbers if x % 2 == 0}
print(result)
```

<p>Generator Comprehension:Generator comprehension (also called a generator expression) creates a generator that produces values one at a time.
<p><b>Syntax:</b>(expression for item in iterable if condition)


```python
numbers = (x*x for x in range(1, 6))
for value in numbers:
    print(value)
```

<p>Explanation:
<p>Parentheses () are used.
<p>Values are generated one at a time.
<p>It is memory efficient for large sequences.
<p>It supports lazy evaluation.


<p><b>Nested List Comprehension</b>
<p>A nested list comprehension contains more than one for clause.
<p>It is useful for working with nested lists and matrices.


```python
matrix = [[1, 2], [3, 4], [5, 6]]
result = [x for row in matrix for x in row]
print(result)
```

<p>Explanation:
<p>The first for loop accesses each row.
<p>The second for loop accesses each element in that row.
<p>All elements are stored in a single list.


```python
#Difference Between Normal Loop and Comprehension
#Using Normal For Loop:
squares = []

for x in range(1, 6):
    squares.append(x*x)

print(squares)
```

```python
#Using List Comprehension:
squares = [x*x for x in range(1, 6)]
print(squares)
```

```python
#Both produce the same output, but list comprehension requires less code.
```

<p><b>Advantages of Comprehensions</b>
<p>1.	Reduces the number of lines of code.
<p>2.	Makes code concise and readable.
<p>3.	Helps create collections easily.
<p>4.	Supports conditional filtering.
<p>5.	Avoids writing lengthy loops for simple transformations.
<p>6.	Generator comprehensions save memory by generating values when needed.
